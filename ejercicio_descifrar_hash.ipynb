Ejercicio 1: hash

```python
import hashlib
from itertools import permutations
```

```python
def brute_force_sha256(objetivo):
    contador = 0
    numero = 0
    limite = 10**10

    while numero < limite:
        intento = f"{numero:010d}"
        resultado = hashlib.sha256(intento.encode()).hexdigest()

        contador += 1

        # Mostrar avance cada cierto número de intentos
        if contador % 1_000_000 == 0:
            print(f"Intentos: {contador}")

        if resultado == objetivo:
            print("\n Secuencia encontrada")
            print(f"Valor: {intento}")
            print(f"Intentos totales: {contador}")
            return intento

        numero += 1

    print("No se encontró ninguna coincidencia")
    return None
```

```python
hash = hashlib.sha256("0123456789".encode()).hexdigest()
print(f"Por ejemplo el hash de '0123456789': {hash}\n")

hash = input("Ingrese el hash a descifrar: ")
brute_force_sha256(hash)
```

Ejercicio 2: arbol merkle

```python
class ArbolMerkle:

    def __init__(self, datos):
        self.datos = datos
        self.raiz = None
        self.niveles = []
        self._construir()

    def _hash(self, valor):
        return hashlib.sha256(str(valor).encode()).hexdigest()

    def _crear_nivel(self, lista_hashes):
        nuevo_nivel = []

        i = 0
        while i < len(lista_hashes):
            if i + 1 < len(lista_hashes):
                combinado = lista_hashes[i] + lista_hashes[i+1]
                nuevo_nivel.append(self._hash(combinado))
            else:
                nuevo_nivel.append(lista_hashes[i])
            i += 2

        return nuevo_nivel

    def _construir(self):
        if not self.datos:
            return

        actual = [self._hash(d) for d in self.datos]
        self.niveles.append(actual)

        while len(actual) > 1:
            actual = self._crear_nivel(actual)
            self.niveles.append(actual)

        self.raiz = actual[0]

    def mostrar(self):
        print(f"Raíz del árbol: {self.raiz}\n")

        total = len(self.niveles)

        for i in range(total - 1, -1, -1):
            espacios = " " * (total - i)
            print(f"{espacios}Nivel {i}: {self.niveles[i]}")


def encontrar_orden(objetivo, lista_tx):
    intentos = 0

    for orden in permutations(lista_tx):
        intentos += 1

        arbol = ArbolMerkle(list(orden))

        if arbol.raiz == objetivo:
            print("\nOrden correcto encontrado:")
            print(list(orden))
            print(f"Intentos: {intentos}")
            return list(orden)

    print("No se encontró un orden válido.")
    return None
```

```python

tx_original = ["trans1", "trans2", "trans3", "trans4", "trans5"]
arbol = ArbolMerkle(tx_original)

arbol.mostrar()

tx_desorden = ["trans3", "trans1", "trans5", "trans2", "trans4"]

encontrar_orden(arbol.raiz, tx_desorden)
```
